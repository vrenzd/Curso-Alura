# <font color='green'>Problema A</font>

Avaliando nosso dataset é possível verificar que a **proporção de homens** como chefes de domicílios é de quase **70%**. Precisamos **selecionar aleatoriamente grupos de 10 indivíduos** para verificar as diferenças entre os rendimentos em cada grupo. Qual a **probabilidade de selecionamos um grupo que apresente a mesma proporção da população**, ou seja, selecionarmos um grupo que seja **composto por 7 homens e 3 mulheres**?

#### <font color='blue'>Como tarefa extra, verifique a real proporção de homens e mulheres em nosso dataset (vimos como fazer isso em nosso primeiro curso de estatística).</font>

#### <font color='red'>Verifique que tipo de distribuição de probabilidade se encaixa neste experimento.</font>

### Solução

```python
#Verificando a proporção de homens e mulheres
sexo_counts = df['Sexo'].value_counts(normalize=True) * 100
sexo_counts
```

```python
#Proporção de homens (sexo = 0)
prop_homens = sexo_counts[0] / 100
round(prop_homens, 2)
```

```python
#Proporção de mulheres (sexo = 1)
prop_mulheres = sexo_counts[1] / 100
round(prop_mulheres, 2)
```

```python
from scipy.special import comb

prob_7_homens = comb(10,7) * (prop_homens**7) * ((1 - prop_homens)**3)
round(prob_7_homens, 4)
```

# <font color='green'>Problema B</font>

Ainda sobre a questão anterior, **quantos grupos de 10 indivíduos** nós precisaríamos selecionar, de forma aleatória, para conseguir **100 grupos compostos por 7 homens e 3 mulheres**?

#### <font color='red'>Lembre-se da forma de cálculo da média de uma distribuição binomial</font>

### Solução

```python
grupos_necessarios = 100 / prob_7_homens
int(round(grupos_necessarios))
```

# <font color='green'>Problema C</font>

Um cliente nos encomendou um estudo para avaliar o **rendimento dos chefes de domicílio no Brasil**. Para isso precisamos realizar uma nova coleta de dados, isto é, uma nova pesquisa de campo. Após reunião com o cliente foi possível elencar o seguinte conjunto de informações:

> A. O resultado da pesquisa precisa estar pronto em **2 meses**;

> B. Teremos somente **R$\$$ 150.000,00** de recursos para realização da pesquisa de campo; e
    
> C. Seria interessante uma **margem de erro não superior a 10% em relação a média estimada**.

Em nossa experiência com estudos deste tipo, sabemos que o **custo médio por indivíduo entrevistado fica em torno de R$\$$ 100,00**. Com este conjunto de fatos avalie e obtenha o seguinte conjunto de informações para passar ao cliente:


> 1. Para obter uma estimativa para os parâmetros da população (renda dos chefes de domicílio no Brasil), realize uma amostragem aleatória simples em nosso conjunto de dados. Essa amostra deve conter 200 elementos (utilize random_state = 101 para garantir que o mesmo experimento posso ser realizado novamente). Obtenha a média e o desvio-padrão dessa amostra.
    
> 2. Para a **margem de erro** especificada pelo cliente obtenha os **tamanhos de amostra** necessários para garantir os **níveis de confiança de 90%, 95% e 99%**.
    
> 3. Obtenha o **custo da pesquisa** para os três níveis de confiança.
    
> 4. Para o maior nível de confiança viável (dentro do orçamento disponível), obtenha um **intervalo de confiança para a média da população**.
    
> 5. Assumindo o **nível de confiança escolhido no item anterior**, qual **margem de erro** pode ser considerada utilizando todo o recurso disponibilizado pelo cliente?
    
> 6. Assumindo um **nível de confiança de 95%**, **quanto a pesquisa custaria ao cliente** caso fosse considerada uma **margem de erro de apenas 5%** em relação a média estimada?


# <font color='blue'>Solução do item 1</font>

### Seleção de uma amostra aleatório simples

#### <font color='red'>Lembre-se de utilizar *random_state = 101*</font>

```python
amostra_renda = df['Renda'].sample(n=200, random_state=101)
```

```python
media_amostra = amostra_renda.mean()
```

```python
desvio_padrao_amostra = amostra_renda.std()
```

### Dados do problema

```python
print(f'Média da amostra: R$ {media_amostra:.2f}')
print(f'Desvio padrão da amostra: R$ {desvio_padrao_amostra:.2f}')
```

# <font color='blue'>Solução do item 2</font>

### Obtenha a margem de erro

#### <font color='red'>Lembre-se que a margem de erro deve estar na mesma unidade da variável que está sendo estudada (R$)</font>

```python
margem_erro_percentual = 0.1
margem_erro_absoluta = media_amostra * margem_erro_percentual
print(f'Margem de erro absoluta: R$ {margem_erro_absoluta:.2f}')
```

### Tamanho da amostra ($1 - \alpha = 90\%$)

```python
confianca_90 = 0.90
z_90 = norm.ppf(0.5 + (confianca_90 / 2))
print(f"Valor Z para 90% de confiança: {z_90:.2f}")
```

```python
tamanho_amostra_90 = ((z_90 * desvio_padrao_amostra) / margem_erro_absoluta)**2
print(f'{tamanho_amostra_90:.0f} elementos')
```

### Tamanho da amostra ($1 - \alpha = 95\%$)

```python
confianca_95 = 0.95
z_95 = norm.ppf(0.5 + (confianca_95 / 2))
print(f'Valor Z para 95% de confiança: {z_95:.2f}')
```

```python
tamanho_amostra_95 = np.ceil(((z_95 * desvio_padrao_amostra) / margem_erro_absoluta)**2)
print(f'{tamanho_amostra_95:.0f} elementos')
```

### Tamanho da amostra ($1 - \alpha = 99\%$)

```python
confianca_99 = 0.99
z_99 = norm.ppf(0.5 + (confianca_99 / 2))
print(f'Valor Z para 99% de confiança: {z_99:.2f}')
```

```python
tamanho_amostra_99 = ((z_99 * desvio_padrao_amostra) / margem_erro_absoluta)**2
print(f'{tamanho_amostra_99:.0f} elementos')
```

# <font color='blue'>Solução do item 3</font>

```python
custo_por_individuo = 100
```

### Custo da pesquisa para o nível de confiança de 90%

```python
custo_90 = tamanho_amostra_90 * custo_por_individuo
print(f'R$ {round(custo_90, 2)}')
```

### Custo da pesquisa para o nível de confiança de 95%

```python
custo_95 = tamanho_amostra_95 * custo_por_individuo
print(f'R$ {round(custo_95, 2)}')
```

### Custo da pesquisa para o nível de confiança de 99%

```python
custo_99 = tamanho_amostra_99 * custo_por_individuo
print(f'R$ {round(custo_99, 2)}')
```

# <font color='blue'>Solução do item 4</font>

```python
margem_erro_95 = z_95 * (desvio_padrao_amostra / np.sqrt(tamanho_amostra_95))

intervalo_inferior_95 = media_amostra - margem_erro_95
intervalo_superior_95 = media_amostra + margem_erro_95

print(f'Intervalo de confiança para a média da população: [R$ {intervalo_inferior_95:.2f}, R$ {intervalo_superior_95:.2f}]')
```

# <font color='blue'>Solução do item 5</font>

```python
orcamento_disponivel = 150000
```

```python
tamanho_amostra_orcamento = orcamento_disponivel / custo_por_individuo
margem_erro_orcamento = z_95 * (desvio_padrao_amostra / np.sqrt(tamanho_amostra_orcamento))
```

```python
print(f'Tamanho da amostra com orçamento disponível: {tamanho_amostra_orcamento:.0f} elementos')
print(f'Margem de erro com orçamento disponível (95% de confiança): R$ {margem_erro_orcamento:.2f}')
```

# <font color='blue'>Solução do item 6</font>

```python
# margem de erro desejada (5% da amostra)
margem_erro = media_amostra * 0.05
margem_erro
```

```python
# Tamanho da amostra para 95% de confiança com 5% de margem de erro
tamanho_amostra_erro = ((z_95 * desvio_padrao_amostra) / margem_erro)**2
tamanho_amostra_erro
```

```python
# Custo da pesquisa
custo = tamanho_amostra_erro * custo_por_individuo
```

```python
print(f"Custo da pesquisa para 95% de confiança e 5% de margem de erro: R$ {custo:,.2f}")
```
